```python
r=int(input("Enter number of rats: "))
unit=int(input("Enter food consumed by each rat: "))
n=int(input("Enter number of houses: "))
arr=list(map(int, input("Enter food in each house: ").split()))

if arr is None:
    print(-1)
else:
    fn= r*unit
    food = 0
    houses = 0
    for i in range(n):
        food = food + arr[i]
        houses = houses + 1

        if food >= fn:
            break

    if food < fn:
        print(0)
    else:
        print(houses)
```

```python
a=int(input("Enter your number:"))
digit=a
rev=0
while a>0:
  num=a%10
  rev=rev*10+num
  a=a//10
if rev==digit:
  print("Number is palindrom")
else:
  print("Number is not palindrom")
```
