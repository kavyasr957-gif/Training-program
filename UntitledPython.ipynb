```python
n=int(input("Enter jar capacity:"))
k=int(input("Enter min candies:"))
t=n
i='y'
ts=0
while i!='n':
  x=int(input("Enter candies to sold:"))
  if x>n:
    print("Invalid input")
  else:
    n=n-x
    ts=ts+x
    print("Sold:",x)
    if n<=k:
      n=t
    print("Available candies:",n)
    print("Total sold",ts)
```
